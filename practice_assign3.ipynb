```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Conv2D,MaxPooling2D,Flatten
```

```python
test_data=pd.read_csv(r'D:\Documents\PICT SEM 8\LP-5\DL\Assign3\fashion-mnist_test.csv')
train_data=pd.read_csv(r'D:\Documents\PICT SEM 8\LP-5\DL\Assign3\fashion-mnist_train.csv')
train_data.head()
```

```python
test_data.head()
```

```python
train_data.isna().sum()
test_data.isna().sum()
```

```python
x_train_data=train_data.iloc[:,1:].values
y_train_data=train_data.iloc[:,0].values
x_test_data=test_data.iloc[:,1:].values
y_test_data=test_data.iloc[:,0].values
x_train_data=x_train_data/255.0
x_test_data=x_test_data/255.0
x_train_data=x_train_data.reshape(-1,28,28,1)
x_test_data=x_test_data.reshape(-1,28,28,1)
```

```python
from tensorflow.keras.utils import to_categorical
y_train_data=to_categorical(y_train_data,10)
y_test_data=to_categorical(y_test_data,10)
```

```python
model=Sequential()
model.add(Conv2D(32,(3,3),activation="relu",input_shape=(28,28,1)))
model.add(MaxPooling2D(2,2))
model.add(Conv2D(64,(3,3),activation="relu",input_shape=(28,28,1)))
model.add(MaxPooling2D(2,2))
```

```python
model.add(Flatten())
model.add(Dense(128,activation="relu"))
model.add(Dense(10,activation="softmax"))
model.compile(optimizer="adam",loss="categorical_crossentropy",metrics=["accuracy"])
```

```python
model.summary()
```

```python
model.fit(x_train_data,y_train_data,epochs=10,validation_split=0.05)
```

```python
model.evaluate(x_test_data,y_test_data)
predicted=model.predict(x_test_data)
for i in range(25):
    print("Predicted Value:",predicted[i].argmax())
    print("Real Value is:",y_test_data[i])
```
