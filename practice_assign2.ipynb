```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Dropout,Flatten
```

```python
data=pd.read_csv("D:\Documents\PICT SEM 8\LP-5\DL\Assign2\IMDB Dataset.csv")
data.head()
```

```python
data.shape
```

```python
data.isna().sum()
```

```python
x=data["review"]
y=data["sentiment"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()
y_train=encoder.fit_transform(y_train)
y_test=encoder.transform(y_test)
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
tokenizer=Tokenizer(num_words=20000)
tokenizer.fit_on_texts(x_train)
x_train_seq=tokenizer.texts_to_sequences(x_train)
x_test_seq=tokenizer.texts_to_sequences(x_test)
x_train_pad=pad_sequences(x_train_seq,maxlen=300)
x_test_pad=pad_sequences(x_test_seq,maxlen=300)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,LSTM,Dropout,Embedding
model=Sequential()
model.add(Embedding(input_dim=20000,output_dim=128,input_length=300))
model.add(LSTM(128,return_sequences=True))
model.add(Dropout(0.5))
model.add(LSTM(64))
model.add(Dropout(0.5))
model.add(Dense(1,activation="sigmoid"))

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
model.summary()
```

```python
model.fit(x_train_pad,y_train,epochs=10,validation_split=0.05)
```

```python
output=model.evaluate(x_test_pad,y_test)
```
